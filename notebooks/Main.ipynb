```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration du style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (20, 6)

# Liste de tes datasets
datasets = [
    "bank77", "clinc_intent", "mtop_intent", "mtop_domain",
    "massive_intent", "massive_domain", "stackex", "arxiv", "reddit",
    "go_emotions", "few_rel_nat", "few_nerd_nat", "few_event", "clinc_domain"
]

# Chemin vers les données standardisées
# (Puisque nous sommes dans /notebooks, on remonte d'un niveau pour accéder à /src)
base_path = "../src/clusterllm/datasets"

for ds in datasets:
    fig, axes = plt.subplots(1, 2)
    fig.suptitle(f"Dataset: {ds.upper()}", fontsize=16, fontweight='bold', y=1.05)
    
    for i, split in enumerate(["large", "small"]):
        file_path = os.path.join(base_path, ds, f"{split}.jsonl")
        
        if os.path.exists(file_path):
            df = pd.read_json(file_path, lines=True)
            counts = df['label'].value_counts()
            
            # On affiche les 25 classes les plus fréquentes
            top_n = counts.head(25)
            sns.barplot(x=top_n.index.astype(str), y=top_n.values, ax=axes[i], palette="magma")
            
            # Calcul du ratio de déséquilibre
            # Ratio = (Min / Max)
            imb_ratio = (counts.min() / counts.max())
            
            axes[i].set_title(f"Split {split.upper()} (Total: {len(df)} | Ratio: {imb_ratio:.2f})")
            axes[i].tick_params(axis='x', rotation=45)
            axes[i].set_ylabel("Nombre d'exemples")
        else:
            axes[i].text(0.5, 0.5, f"{split}.jsonl non trouvé", ha='center', va='center')
            axes[i].axis('off')

    plt.tight_layout()
    plt.show()
    print("-" * 100)
```

```python
stats = []
for ds in datasets:
    for split in ["large", "small"]:
        path = os.path.join(base_path, ds, f"{split}.jsonl")
        if os.path.exists(path):
            df = pd.read_json(path, lines=True)
            counts = df['label'].value_counts()
            stats.append({
                'dataset': ds,
                'split': split,
                'ratio': counts.min() / counts.max(),
                'nb_classes': len(counts),
                'total_rows': len(df)
            })

# Création d'un tableau de bord
df_stats = pd.DataFrame(stats).sort_values(by='ratio')
print("---  DES DATASETS LES PLUS DÉSÉQUILIBRÉS ---")
print(df_stats.head(30))
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration du style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (20, 6)

# Liste mise à jour avec les noms EXACTS des dossiers de la capture d'écran
datasets = [
    "arxiv_fine",
    "banking77",
    "clinc",
    "clinc_domain",
    "few_event",
    "few_nerd_nat",
    "few_rel_nat",
    "go_emotion",
    "massive_intent",
    "massive_scenario",
    "mtop_domain",
    "mtop_intent",
    "reddit",
    "stackexchange"
]

# Chemin vers les données (Vérifie que ce chemin pointe bien vers le dossier racine "datasets" de ta capture)
base_path = "../datasets"

for ds in datasets:
    fig, axes = plt.subplots(1, 2)
    fig.suptitle(f"Dataset: {ds.upper()}", fontsize=16, fontweight='bold', y=1.05)
    
    for i, split in enumerate(["large", "small"]):
        file_path = os.path.join(base_path, ds, f"{split}.jsonl")
        
        if os.path.exists(file_path):
            df = pd.read_json(file_path, lines=True)
            
            # Vérifie que la colonne s'appelle bien 'label' (parfois c'est 'intent' ou 'category')
            if 'label' in df.columns:
                counts = df['label'].value_counts()
                
                # On affiche les 25 classes les plus fréquentes
                top_n = counts.head(25)
                sns.barplot(x=top_n.index.astype(str), y=top_n.values, ax=axes[i], palette="magma")
                
                # Calcul du ratio de déséquilibre (Min / Max)
                imb_ratio = (counts.min() / counts.max())
                
                axes[i].set_title(f"Split {split.upper()} (Total: {len(df)} | Ratio: {imb_ratio:.2f})")
                axes[i].tick_params(axis='x', rotation=45)
                axes[i].set_ylabel("Nombre d'exemples")
            else:
                axes[i].text(0.5, 0.5, f"Erreur : pas de colonne 'label'", ha='center', va='center', color='red')
                axes[i].axis('off')
        else:
            axes[i].text(0.5, 0.5, f"{split}.jsonl non trouvé\n{file_path}", ha='center', va='center')
            axes[i].axis('off')

    plt.tight_layout()
    plt.show()
    print("-" * 100)
```

```python
stats = []
for ds in datasets:
    for split in ["large", "small"]:
        path = os.path.join(base_path, ds, f"{split}.jsonl")
        if os.path.exists(path):
            df = pd.read_json(path, lines=True)
            counts = df['label'].value_counts()
            stats.append({
                'dataset': ds,
                'split': split,
                'ratio': counts.min() / counts.max(),
                'nb_classes': len(counts),
                'total_rows': len(df)
            })

# Création d'un tableau de bord
df_stats = pd.DataFrame(stats).sort_values(by='ratio')
print("---  DES DATASETS LES PLUS DÉSÉQUILIBRÉS ---")
print(df_stats.head(30))
```
